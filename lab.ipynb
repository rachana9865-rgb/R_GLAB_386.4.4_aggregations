## R_GLAB_386.4.4 Aggregations

```python
import pandas as pd
import numpy as np
```

```python
data = {
    'Category': ['A', 'A', 'B', 'B', 'A', 'B','A','B'],
    'Value': [10, 15, 20, 25, 30, 35,40,50],
    'Fee' :[20000,25000,26000,22000,24000,35000,4000,5600],
    'Duration':['30day','40days','35days','40days','60days','60days','50days','60days'],
    'Discount':[1000,2300,1200,2500,2000,2000,500,600]
}

df = pd.DataFrame(data)

print(df.head())
```

```python
# Calculate the total sum of the value column
total_sum = df['Value'].agg('sum')

print(f'Total Sum: {total_sum}')

# Alternative

total_sum1 = df[['Value']].sum()

print(total_sum1)
```

```python
df['Value'].agg('cumsum')
```

```python
df[['Value']].cumsum()
```

```python
# Do both the agg and alternative to find: mean
mean1 = df['Value'].agg('mean')
print(f'Mean: {mean1}')

mean2 = df[['Value']].mean()
print(mean2)
```

```python
# Do both the agg and alternative to find: max
max1 = df['Value'].agg('max')
print(max1)

max2 = df[['Value']].max()
print(max2)
```

```python
# Do both the agg and alternative to find: count
count1 = df['Value'].agg('count')
print(count1)

count2 = df[['Value']].count()
print(count2)
```

## Multiple Aggregations

```python
result = df[['Fee', 'Discount']].agg('sum')

result
```

```python
result = df[['Fee', 'Discount']].sum()

result
```

```python
result = df[['Fee', 'Discount']].agg(['min', 'mean', 'max'])

result
```

```python
result_group = df.groupby('Category')[['Value', 'Fee', 'Discount']].agg('sum')

result_group
```
